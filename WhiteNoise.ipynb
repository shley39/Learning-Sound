```python
import numpy as np
import sounddevice as sd
#importing a necessary library

def main():
    mysound = white_noise()
    sd.play(mysound)
    #Changing the number in to sounds. Using 40,000 samples in less than a second. The numbers are moving the position of the speaker membrane. Creating a sound waveform.
    #Sound is a vibration of air, so moving the speaker membrane and vivrating the air creates the sound
    sd.wait()
    #Waiting until the sound is finished



def white_noise(
        duration: float=1.0,
        amplitude: float=0.5,
        sample_rate: int=44100
        #This is all variable
        #Why 44100? Human ear can only hear up to 20,000 Hz but we need the up and down so times 2 and becomes 44100 (adding 100 because of the CD standard)
        ) -> np.ndarray:
        #This means that this is all numbers from Numpy, something like a list of numbers but more efficient.
    """
    Generate white noise.
    """
    #Calculate the number of samples based on duration and sample rate
    n_samples = int(duration * sample_rate)
    #creating tons of samples
    #Generate white noise using a uniform distribution
    noise = np.random.uniform(-1, 1, n_samples)
    #Uniform means generating random numbers. Generating n_samples random numbers in the range from -1 to 1 but in uniform distribution.

    noise *= amplitude  # Scale the noise by the amplitude
    #Adjusting the volume, since amplitude is 0.5, making it half smaller. Making each sample half.
    #this Def can only make the numbers of the sound not the sound itself. 
    return noise

main()
```
